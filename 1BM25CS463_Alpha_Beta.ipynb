```python


def get_leaves(node):
    if not isinstance(node, list):
        return [node]

    result = []
    for child in node:
        result.extend(get_leaves(child))
    return result


def alpha_beta(node, is_max, alpha, beta, pruned):
    if not isinstance(node, list):
        print("Evaluating leaf:", node)
        return node

    if is_max:
        best = float('-inf')

        for i, child in enumerate(node):
            value = alpha_beta(child, False, alpha, beta, pruned)
            best = max(best, value)
            alpha = max(alpha, best)

            if alpha >= beta:
                for remaining in node[i + 1:]:
                    pruned.extend(get_leaves(remaining))
                print("Pruned nodes:", pruned)
                break

        return best

    else:
        best = float('inf')

        for i, child in enumerate(node):
            value = alpha_beta(child, True, alpha, beta, pruned)
            best = min(best, value)
            beta = min(beta, best)

            if alpha >= beta:
                for remaining in node[i + 1:]:
                    pruned.extend(get_leaves(remaining))
                print("Pruned nodes:", pruned)
                break

        return best


print("Enter 8 leaf node values from left to right:")

values = []
for i in range(8):
    value = int(input("Enter value " + str(i + 1) + ": "))
    values.append(value)

tree = [
    [[values[0], values[1]], [values[2], values[3]]],
    [[values[4], values[5]], [values[6], values[7]]]
]

pruned = []

print("\nAlpha-Beta Pruning Process:")

result = alpha_beta(
    tree, True,
    float('-inf'),
    float('inf'),
    pruned
)

print("\n----- FINAL RESULT -----")
print("Root MAX Value:", result)
print("Pruned Nodes:", pruned)
print("Optimal Value:", result)
```
