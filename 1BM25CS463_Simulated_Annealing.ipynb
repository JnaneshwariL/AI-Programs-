```python

import random
import math

def cost(x):
    return x * x

current = int(input("Enter initial state: "))
T = float(input("Enter initial temperature: "))
cooling_rate = float(input("Enter cooling rate (0 to 1): "))
min_temperature = float(input("Enter minimum temperature: "))

random.seed(7)

steps = 0

while T > min_temperature:

    next_state = current + random.choice([-1, 1])

    delta_E = cost(current) - cost(next_state)

    if delta_E > 0:
        current = next_state

    else:
        probability = math.exp(delta_E / T)

        if random.random() < probability:
            current = next_state

    T = T * cooling_rate
    steps += 1

print("\n----- SIMULATED ANNEALING RESULT -----")
print("Final State:", current)
print("Minimum Cost:", cost(current))
print("Total Iterations:", steps)
```
