```python


n = int(input("Enter number of facts: "))

KB = set()

for i in range(n):
    fact = input("Enter fact: ")
    KB.add(fact)

r = int(input("\nEnter number of rules: "))

rules = []

for i in range(r):
    condition = input("Enter condition: ")
    conclusion = input("Enter conclusion: ")
    rules.append((condition, conclusion))

query = input("\nEnter query: ")

while True:
    new = set()

    for condition, conclusion in rules:
        if condition in KB and conclusion not in KB:
            new.add(conclusion)

    if not new:
        break

    print("\nNew facts generated:", new)

    KB.update(new)

    if query in KB:
        break

print("\nFinal Knowledge Base:")

for fact in sorted(KB):
    print(fact)

if query in KB:
    print("\nQuery proved!")
    print("Result: TRUE")
else:
    print("\nQuery cannot be proved!")
    print("Result: FALSE")
```
